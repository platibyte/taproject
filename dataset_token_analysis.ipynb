```python
import pandas as pd
import nltk

df = pd.read_parquet('dataset.parquet')
```

```python
df['Answer'].isna().value_counts()
df = df.dropna()
df['Answer'].isna().value_counts()
```

```python
def tokenize(x):
    tokens = nltk.word_tokenize(x)
    return len(tokens)
```

```python
df['lens'] = df['Answer'].apply(lambda x: tokenize(x))
df['lens'].median()
```
